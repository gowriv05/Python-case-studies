Case Study– Departmental Data Tracker
Instructions:

Create a dictionary department with keys: "dept_name", "faculty", "courses", "budget".

Add and modify items using dictionary methods (update, pop, get, etc.).

Create two sets: online_courses and offline_courses.

Perform union, intersection, difference, and symmetric_difference.
Create a frozenset named permanent_faculty.

Display all results

```python
# Departmental Data Tracker
# Step 1: Department Dictionary
department={
            "dept_name":"IT",
            "faculty":["Raj","Mohan","Sunitha"],
            "courses":["Excel","Powerbi","Python"],
            "budget":100000}
print("Department:",department)

# Step 2: Modify values
# add
department.update({"location":"C Block"})
# modify
department.update({"budget":120000})
# get
print("\nDepartment name:",department.get("dept_name"))
print("\nBudget:",department.get("budget"))
print("\nDepartment:",department)
# remove an item using pop()
removed_item=department.pop("location")
print("\nRemoved item:",removed_item)
print("\nUpdated department:",department)

# Step 3: Set Operations
online_courses={"Excel","Python","Powerbi","SQL"}
offline_courses={"Excel","SQL","Tableu","Statistics"}
# union
print("\nUnion:")
print(online_courses.union(offline_courses))
# intersection
print("\nIntersectio:")
print(online_courses.intersection(offline_courses))
# difference
print("\nOnline-Offline difference:")
print(online_courses.difference(offline_courses))
print("\nOffline-Online difference:")
print(offline_courses.difference(online_courses))
# symmetric difference
print("\nSymmetric difference:")
print(offline_courses.symmetric_difference(online_courses))

# Step 4: Frozenset Example
permanent_faculty=frozenset(["Priya","Ravi","Sandhiya"])
print("\nPermanent Faculty:",permanent_faculty)
```
