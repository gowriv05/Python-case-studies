```python
# Reatail Inventory Manager
# Fixed product categories
categories=("Electronics","Stationary","Clothing")

# Product list
products=["Laptop","Pen","Pant"]

# Ask the user to enter a new product and add it to the list
new_product=input("Enter a new product:")

# Add new product
products.append(new_product)

# Display products
print("Products:",products)

# Ask user for index to remove
index=int(input("\nEnter the index of the product to remove:"))

# Remove product using index
products.pop(index)

# Display final list
print("\nFinal products:",products)
print("Product categories:",categories)
```
