```python
# Contact Directory
# Create a 2D LIST
contacts=[
    ["Hari","7865432221","hari@gmail.com"],
    ["Charu","8765309867","charu@gmail.com"],
    ["Aravind","9876453219","aravind@gmail.com"]
    ]

# Get new contact from user
name=input("Enter name:")
phone=input("Enter phone number:")
email=input("Enter email:")
print("\nContacts:","\n",contacts)

# Add new contact
contacts.append([name,phone,email])
print("\nContacts:","\n",contacts)

# Remove the last item
contacts.pop()

# Display contacts
print("\nContacts:","\n",contacts)

# Sort the list in ascending order based on name
contacts.sort()

print("\nSorted contacts:","\n",contacts)

```
